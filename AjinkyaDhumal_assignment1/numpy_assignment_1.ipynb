```python
# 33 Multiply the hours array by the billing rate array.


df["hours_logged"] = pd.to_numeric(df["hours_logged"], errors="coerce")


df["Billing Rate (USD)"] = pd.to_numeric(
    df["Billing Rate (USD)"].astype(str)
    .str.replace(r"[$,]|USD", "", regex=True)
    .str.strip(),
    errors="coerce"
)


valid = df[["hours_logged", "Billing Rate (USD)"]].dropna()


hours = valid["hours_logged"].to_numpy()
rates = valid["Billing Rate (USD)"].to_numpy()


print(hours * rates)
```

```python
# 34 Check whether the two arrays have the same length.
print(len(hours) == len(rates))
print(hours.shape == rates.shape)
```
